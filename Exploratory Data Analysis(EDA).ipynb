```python
#Exploratory Data Analysis
#Patient Age Trend
import matplotlib.pyplot as plt
import seaborn as sns
print(" Age Summary")
print(df["Age"].describe())
plt.figure(figsize=(8, 4))
sns.histplot(df["Age"], kde=True, color="skyblue", bins=20)
plt.title("Distribution of Patient Age")
plt.xlabel("Age (Years)")
plt.ylabel("Patient Count")
plt.show()
```

```python
# Gender Trend
print("Gender Counts")
print(df["Gender"].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Gender", hue="Gender", palette="Set2", legend=False)
plt.title("Patient Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
#Department Wise Trend
print("Department Breakdown")
print(df["Department_Patient"].value_counts())

plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    y="Department_Patient",
    hue="Department_Patient",
    order=df["Department_Patient"].value_counts().index,
    palette="viridis",
)
plt.title("Patient Admissions by Department")
plt.xlabel("Count")
plt.ylabel("Department")
plt.show()
```

```python
#Diagnosis Trend
print("Diagnosis Counts")
print(df["Diagnosis"].value_counts())

plt.figure(figsize=(9, 5))
sns.countplot(
    data=df,
    y="Diagnosis",
    hue="Diagnosis",
    order=df["Diagnosis"].value_counts().index,
    palette="Spectral",
)
plt.title("Most Common Diagnoses")
plt.xlabel("Count")
plt.ylabel("Diagnosis")
plt.show()
```

```python
# Numerical Columns Trend
import matplotlib.pyplot as plt
import seaborn as sns
numeric_cols = ['Length_of_Stay_Days', 'Wait_Time_Minutes', 'Treatment_Cost_USD']
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True, bins=20)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# Severity Levels Trend
plt.figure(figsize=(6,4))
print(df["Severity_Level"].value_counts())
sns.countplot(x='Severity_Level', data=df, order=df['Severity_Level'].value_counts().index)
plt.title("Distribution of Severity Levels")
plt.show()
```

```python
# Patients by City Trend
plt.figure(figsize=(10,6))
print(df["City"].value_counts())
sns.countplot(y='City', data=df, order=df['City'].value_counts().index)
plt.title("Patients by City")
plt.show()
```

```python
#Treatment Outcomes Trend
plt.figure(figsize=(6,4))
print(df["Outcome"].value_counts())
sns.countplot(x='Outcome', data=df)
plt.title("Treatment Outcomes")
plt.show()
```
